```python
#1. find the total sales/revenue
import matplotlib.pyplot as plt
```

PART C- Business requirements

```python
#1.
sales_category = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)

print(sales_category)
print("Highest Sales Category:", sales_category.idxmax())
```

```python
#2. country with most and worst performance
region_sales = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)

print(region_sales)
print("Best Region:", region_sales.idxmax())
print("Worst Region:", region_sales.idxmin())
```

```python
#3. customer segment with most contribution
customer_segment = df.groupby("Customer_Segment")["Sales"].sum().sort_values(ascending=False)

print(customer_segment)
print("Top Customer Segment:", customer_segment.idxmax())
```

```python
#4. category with highest profit
```

```python
#7. Outlier analysis
outlier_analysis = df.groupby("Payment_Method")["Customer_Segment"].sum().sort_values(ascending=False)

print(outlier_analysis)
print("Unusual transactions:", outlier_analysis.idxmax())
```
