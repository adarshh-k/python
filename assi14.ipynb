```python
import pandas as pd
data={
    "Name":['Rahul','PRIYA','arjun','Sneha','rahul','Akhil',None,'Meera','PRIYA'],
    'Age':['22','23',None,'21','22','24','25','23','23'],
    'City':['kochi','Kochi','KOCHI','Trivandrum',None,'kochi','Calicut','CALICUT','Kochi'],
    'Course':['Data Science','data science','DATA SCIENCE','Python',None,'Python','Data Science','python','data science'],
    'Salary':['25000','30000','28000',None,'25000','32000','35000','30000','30000'],
    'Join_Date':['2025-01-10','15/01/2025','2025-02-05',None,'2025-01-10','20/02/2025','2025-03-01','invalid','15/01/2025']
    }
df=pd.DataFrame(data)
df
```

```python
#1. string cleaning
df['Name']=df['Name'].str.strip().str.title()
df['City']=df['City'].str.strip().str.title()
df['Course']=df['Course'].str.strip().str.title()
```

```python
df
```

```python
#2. finding no.of missing values in each columns
df.isnull().sum()
```

```python
#filling missed age and salary with using appropriate statistical value
#Here age and salary are in string format so to fill these 2 first convert it to numeric
df['Age']=pd.to_numeric(df['Age'])
```

```python
df['Salary']=pd.to_numeric(df['Salary'])
```

```python
df['Age']=df['Age'].fillna(df['Age'].mean())
```

```python
df['Salary']=df['Salary'].fillna(df['Salary'].mean())
```

```python
df
```

```python
#filling missing city with 'unknown'
df['City']=df['City'].fillna('Unknown')
```

```python
df
```

```python
#filling missing course with 'Not Specified'
df['Course']=df['Course'].fillna('Not Specified')
```

```python
df
```

```python
#3. find duplicate rows
df.duplicated()
```

```python
#counting no.of duplicated rows
df.duplicated().sum()
```

```python
#removing all duplicate rows
df.drop_duplicates(inplace=True)
df
```

```python
#display cleaned dataframe
df
```

```python
#4. datatype conversion
df['Age']=pd.to_numeric(df['Age'])
```

```python
df['Salary']=pd.to_numeric(df['Salary'])
```

```python
df.dtypes
```

```python
# Handling the invalid date without any error
df['Join_Date']=pd.to_datetime(df['Join_Date'],errors='coerce',dayfirst=True)
# This is used because error='coerce converts invalid dates to 'NaT' or 'NaN' which is useful before filling the missing values.
```

```python
df
```

```python
df['Join_Date']=df['Join_Date'].fillna(df['Join_Date'].mode()[0])
```

```python
df
```

```python
#5. complete data cleaning
#inspect the data
df.info()
```

```python
#clean string values
df['Name']=df['Name'].fillna('Unknown').str.strip().str.title()
df['City']=df['City'].fillna('Unknown').str.strip().str.title()
df['Course']=df['Course'].fillna('Not Specified').str.strip().str.title()
```
