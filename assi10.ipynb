```python
#1. class student
class Student:
    def __init__(a,name,age):
        a.name=name
        a.age=age
s1=Student('Rahul',22)
print('Name:',s1.name)
print('Age:',s1.age)
```

```python
#2. car attributes
class Car:
    def __init__(self,brand,model):
        self.brand=brand
        self.model=model
c1=Car('Toyota','innova')
print('Brand:',c1.brand)
print('Model:',c1.model)
        
```

```python
#3. employeee
class Employee:
    def __init__(self,name,department,salary):
        self.name=name
        self.department=department
        self.salary=salary
e1=Employee('Anu','IT',30000)
print("Name:",e1.name)
print('Department:',e1.department)  
print('Salary:',e1.salary)      
```

```python
#4. bank account
class BankAccount:
    def __init__(self,account_holder,balance):
        self.account_holder=account_holder
        self.balance=balance
    def func(a):
        print('Account Holder:',a.account_holder)
        print('Balance:',a.balance)
b1=BankAccount('Arun',25000)
b1.func()  
print() 
b2=BankAccount('Rahul',40000)
b2.func()
print()
b3=BankAccount('Neha',35000)
b3.func()
print()
```

```python
#5. rectangle
class Rectangle:
    def __init__(self,length,width):
        self.length=length
        self.width=width
    def area(a):
     result=a.length* a.width
     return result
a1=Rectangle(10,5)
print('Area:',a1.area())

 
```

```python
#6. division by 0
try:
    a=int(input('Enter numerator:'))
    b=int(input('Enter denominator:'))
    print(a/b)
except ZeroDivisionError:
    print('Cannot divided by zeroo')
except NameError:
    print('exception occured in nameerrror')
except ValueError:
    print('Exception occured in valueerror')
except:
    print('an exception occured')
else:
    print('x printed')
finally:
    print('successfully completed')
```

```python
#7. handle invalid input
try:
    a=int(input('Enter numerator:'))
    b=int(input('Enter denominator:'))
    print(a/b)
except ZeroDivisionError:
    print('Cannot divided by zeroo')
except NameError:
    print('exception occured in nameerrror')
except ValueError:
    print('invalid input!! Please enter a number')
except:
    print('an exception occured')
else:
    print('x printed')
finally:
    print('successfully completed')
```

```python
#8. multiple exceptions 
try:
    a=int(input('Enter numerator:'))
    b=int(input('Enter denominator:'))
    print(a/b)
except ZeroDivisionError:
    print('Cannot divided by zeroo')
except ValueError:
    print('invalid input!! Please enter a number')
```

```python
#9. list index exception
numbers=[10,20,30,40,50]
try:
    index=int(input('Enter the index:'))
    print('Element at this index is',numbers[index])
except IndexError:
    print('Invalid index')
except ValueError:
    print('ivalid input!')
```
