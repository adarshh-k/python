```python
import numpy as np
```

```python
import pandas as pd
data={
    'Name':['Anu','Rahul','Akhil','Meera','Neha','Arjun'],
    'Age':[23,np.nan,25,24,np.nan,27],
    'City':['Kochi','Calicut',np.nan,'Kochi','Thrissur','Kochi'],
    'Salary':[25000,30000,35000,np.nan,28000,40000],
    'Department':['IT','HR','IT','Finance','HR',np.nan]
}
```

```python
#create the above dataframe
#display the dataframe
df=pd.DataFrame(data)
df
```

```python
#check the number of null value in each column
df.isna().sum()
```

```python
#display rows containing null values
df[df.isna().any(axis=1)]
```

```python
#replace null value in age with mean age
df['Age']=df['Age'].fillna(df['Age'].mean())
df
```

```python
#replace null value in city with 'unknown'
df['City']=df['City'].fillna('Unknown')
df
```

```python
#replace null value in salary with the median salary
df['Salary']=df['Salary'].fillna(df['Salary'].median())
df
```

```python
#replace null value in department with 'not assigned'
df['Department']=df['Department'].fillna('Not Assigned')
df
```

```python
#checking null values again
df.isna().sum()
```

```python
#final cleared dataframe
df
```
