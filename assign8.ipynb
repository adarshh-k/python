```python
#1. calculating avg of marks
def calculate_average(s1,s2,s3,s4,s5):
    average=(s1+s2+s3+s4+s5)/5
    return average
s1=int(input("Enter mark of subject1:"))
s2=int(input("Enter mark of subject2:"))
s3=int(input("Enter mark of subject3:"))
s4=int(input("Enter mark of subject4:"))
s5=int(input("Enter mark of subject5:"))
avg=calculate_average(s1,s2,s3,s4,s5)
print(f'average is :{avg}')
if avg>=40:
    print('student passed')
else:
    print('student failed!!!!')
```

```python
#2. odd or even
def check_number(num):
    print('even') if num%2==0 else print('odd')
```

```python
check_number(3)
check_number(2)
check_number(9)
check_number(1)
check_number(13)
```

```python
#3. shopping bill
def calculate_bill(price,quantity):
    total=price*quantity
    if total>5000:
        discound=total*10/100
        total=total-discound
        return total
price=int(input('Enter the price:'))
quantity=int(input('Enter the quantity:'))
final=calculate_bill(price,quantity)
print('final amount=',final)

```

```python
#4. largest number
def find_largest(n1,n2,n3):
    return n1 if n1>n2 and n1>n3 else n2 if n2>n1 and n2>n3 else n3
```

```python
find_largest(1,2,3)
```
