```python
#SECTION B
#1.
marks={
    'Anu':85,
    'Rahul':72,
    'Akhil':91,
    'Meera':65,
    'Neha':78
}
```

```python
#student with highest mark
highmark=max(marks,key=marks.get)
print(f'student with highest mark:{highmark}')
```

```python
#student with low mark
lowmark=min(marks,key=marks.get)
print(f'student with lowest mark:{lowmark}')
```

```python
#to find average of marks
totalmark=sum(marks.values())
avg=totalmark/len(marks)
print('Average mark',avg)
```

```python
#creating new dictionary of students who scored more than 75 or more
new={}
for name,mark in marks.items():
    if mark>=75:
        new[name]=mark
print(new)
```

```python
#2. pattern printing
for i in range(6):
    for j in range(1,i+1):
        print(j,end='')
    print()
```

```python
for i in range(5,0,-1):
    for j in range(1,i+1):
        print(j,end='')
    print()
```

```python
#SECTION C

#ATM withdrawal system:
balance=10000
amount=int(input('Enter the amount to withdraw:'))
if amount<=0:
    print('invalid amounttt')
if amount>balance:
    print('amount is greater than balance!!')
else:
    print('withdrawal successfully')
    balance=balance-amount
    print('Remaining balance:',balance)


```
