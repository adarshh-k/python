```python
#1. odd or even using shorthandif
n=int(input("Enter a number:"))
print('even') if n%2==0 else print('odd')
```

```python
#2. displaying largest no using shorthandif
a=input("Enter number1:")
b=input('Enter number2:')
print('a is largest',a) if a>b else print('b is largest',b) 
```

```python
#3. passed of failed in exam
n=int(input('Enter mark:'))
print('Passed') if n>=40 else print('Failed!!')
```

```python
#4. positive or negative number
a=int(input('Enter a number:'))
print('Positive') if a>0 else print('Negative')
```

```python
#5. check customer is eligible for discound
n=int(input('Enter the amount:'))
print('Eliggible for discound') if n>=5000 else print(' Not eligible for discound')
```

```python
#6. day of week using match parent
day=int(input('Enter a day:'))
match day:
    case 1:
      print('Monday')
    case 2:
      print('Tuesday')
    case 3:
      print('Wednesday')
    case 4:
      print('Thursday')
    case 5:
      print('Friday')
    case 6:
      print('Saturday')
    case 7:
      print('Sunday')
```

```python
#7. calculator 
a=int(input('Enter number1:'))
b=int(input('Enter number2:'))
op=input('Enter operator(+-*/)')
match op:
    case '+':
        print(a+b)
    case '-':
        print(a-b)
    case '*':
        print(a*b)
    case '/':
        print(a/b)
```

```python
#8. grade description
grade=input('enter your grade A,B,C,D,F:')
match grade:
    case 'A':
        print('Brilliant')
    case 'B':
        print('Good')
    case 'C':
        print('Average')
    case 'D':
        print('Poor')
    case 'F':
        print('Failed')
    case _:
        print('Not found')
```

```python
#9. traffic signal
n=input('Enter signal (red,yellow,green)')
match n:
    case 'red':
        print('Stop')
    case 'yellow':
        print('Go slowly')
    case 'green':
        print('Ready to go')
    case _:
        print('invalid')
```
