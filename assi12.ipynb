```python
#1. create an array
import numpy as np
x=np.arange(1,11)
print(x)
```

```python
#2. array operations
a=np.array([10,20,30,40,50])
b=np.array([1,2,3,4,5])
print(f'Addition={a+b}')
print(f'Subtraction={a-b}')
print(f'Multiplication={a*b}')
print(f'Division={a/b}')
```

```python
#3. find maximum and minimum
marks=np.array([85,72,91,65,78,88])
print('Maximum mark=',np.max(marks))
```

```python
print('Minimum mark=',np.min(marks))
```

```python
print('Average mark=',np.average(marks))
```

```python
#4. even numbers
import numpy as np
a=np.arange(1,21)
x=np.where(a%2==0)
for i in x:
    for j in i:
        print(a[j])
```

```python
#5. reshape an array
a=np.arange(1,13)
b=a.reshape(3,4)
print(b)
```
