```python
#1. sorting in ascending
numbers=[45,2,78,3,29,56,8]
numbers.sort()
```

```python
numbers
```

```python
#2. sorting by string length
words=['python','ai','machine','data','learning','ml']
words.sort(key=lambda x:len(x))
```

```python
words
```

```python
#3. sorting in descending
numbers=[15,42,7,89,23,64,31]
numbers.sort(reverse=True)
```

```python
numbers
```

```python
#4. filtering even numbers
numbers=[12,7,25,40,33,18,9,50]
even=list(filter(lambda x:x%2==0,numbers))
```

```python
even
```

```python
#5. filtering number>50
numbers=[25,67,12,89,45,73,34,91]
num=list(filter(lambda x:x>50,numbers))
```

```python
num
```

```python
#6. mapping square numbers
numbers=[2,4,6,8,10]
square=list(map(lambda x:x**2,numbers))
```

```python
square
```

```python
#7. mapping numbers to positive
numbers=[-10,5,-7,12,-3,8]
posi=list(map(lambda x:abs(x),numbers))
```

```python
posi
```

```python
#8. reducing
from functools import reduce
num=[10,20,30,40,50]
sum=reduce(lambda a,b:a+b,num) 
```

```python
sum
```

```python
#9. using reduce find largest number
from functools import reduce
numbers=[45,12,89,34,67,23]
great=reduce(lambda a,b:a if a>b else b,numbers)
```

```python
great
```
