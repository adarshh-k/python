```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
df=pd.read_excel(r"C:\Users\MSI MODERN 14\Downloads\01388e40-af68-4b58-aad1-02d9c8248fb4.xlsx")
```

```python
df
```

```python
#sales by category
df.groupby('Category')['Sales'].sum().plot(kind='bar')
plt.xlabel('Category')
plt.ylabel('Sales')
plt.title('Sales by category')
plt.show()
```

. Electronics is the most saled category

```python
#sales by region
df.groupby('Region')['Sales'].sum().plot(kind='bar')
plt.xlabel('Region')
plt.ylabel('Sales')
plt.title('Sales by region')
plt.show()
```

. South is the most saled region

```python
#sales by Customer segment
df.groupby('Customer_Segment')['Sales'].sum().plot(kind='pie',autopct='%.1f%%')
plt.title('Sales by customer_segment')
plt.show()

```

. Small business has the most sales contribution 42.5%

```python
#Monthly sales trend
df.groupby('Month')['Sales'].sum().plot(kind='line')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.title('Sales by month')
plt.show()
```

```python
#Monthly profit trend
df.groupby('Month')['Profit'].sum().plot(kind='line')
plt.xlabel('Month')
plt.ylabel('Profit')
plt.title('Profit by month')
plt.show()
```

```python
#top 5 product of sales
top_5=df.groupby('Product')['Sales'].sum().sort_values(ascending=False).head(5).plot(kind='barh')
#top_5=product_sales.sort_values(ascending=False).head(5)
top_5.plot(kind='barh')
plt.xlabel('Product')
plt.ylabel('Total sales')
plt.title('Top 5 sales product')
plt.show()
```

```python
#sales by city
df.groupby('City')['Sales'].sum().plot(kind='bar')
plt.xlabel('City')
plt.ylabel('Sales')
plt.title('Sales by cityy')
```

. Pune is the city with highest sales

```python
#Quantity vs sales
plt.scatter(df['Quantity'], df['Sales'])
plt.title('Sales vs Quantity')
plt.xlabel('Quantity')
plt.ylabel('Sales')
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
#sales by category and region
sales_table=df.pivot_table(
    index='Category',
    columns='Region',
    values='Sales',
    aggfunc='sum'
)
sns.heatmap(sales_table,annot=True,cmap='coolwarm')
plt.xlabel('Region')
plt.ylabel('Category')
plt.title('sales by category and region')
plt.show()
```

```python
#sales distribution
plt.hist(df['Sales'],bins=30)
plt.title('Histogram')
plt.xlabel('sales')
plt.ylabel('frequency')
plt.show()
```
